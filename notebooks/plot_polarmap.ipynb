```python
import asdf
import matplotlib.pyplot as plt
import matplotlib
```

```python
datadir = "/Users/clowder/data/polar/"
cr = 2122
wave = 193

data = asdf.load(f"{datadir}cr{cr}-193A.asdf")
```

```python
fig, ax = plt.subplots(figsize=(9.5, 7.5), subplot_kw={'projection':data["map_np"].wcs})

im = ax.imshow(data["map_np"].data, cmap=matplotlib.colormaps[f"sdoaia{wave}"], vmin=0, vmax=250)

ax.set_facecolor('black')
# ax.coords.grid(color='white', alpha=.25, ls='dotted')
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title(f"North Pole - SDO/AIA - {wave}$\\AA$ - CR {cr}")
fig.colorbar(im, ax=ax, label='min(DN/s)', extend='max')
```

```python
fig, ax = plt.subplots(figsize=(9.5, 7.5), subplot_kw={'projection':data["map_sp"].wcs})

im = ax.imshow(data["map_sp"].data, cmap=matplotlib.colormaps[f"sdoaia{wave}"], vmin=0, vmax=250)

ax.set_facecolor('black')
# ax.coords.grid(color='white', alpha=.25, ls='dotted')
ax.set_xlabel("")
ax.set_ylabel("")
ax.set_title(f"South Pole - SDO/AIA - {wave}$\\AA$ - CR {cr}")
fig.colorbar(im, ax=ax, label='min(DN/s)', extend='max')
```
